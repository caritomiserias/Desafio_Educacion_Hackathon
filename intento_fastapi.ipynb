```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document # Import Document class

# 1. Define a list of strings for the mini knowledge base
mini_kb_documents_raw = [
    "The capital of France is Paris.",
    "Mount Everest is the highest mountain in the world.",
    "The Pacific Ocean is the largest ocean on Earth.",
    "The Amazon River is the longest river by discharge volume.",
    "The chemical symbol for water is H2O."
]

# Convert raw strings to Document objects
mini_kb_documents = [Document(page_content=doc) for doc in mini_kb_documents_raw]

# 2 & 3. Instantiate OpenAIEmbeddings
embeddings = OpenAIEmbeddings()

# 4 & 5. Create a Chroma vector store from the documents
vector_store = Chroma.from_documents(mini_kb_documents, embeddings)

print("Mini knowledge base documents defined.")
print("OpenAIEmbeddings initialized.")
print("Chroma vector store created.")
```

## Define LangChain RAG Chain

### Subtask:
Set up the Retrieval-Augmented Generation (RAG) chain using LangChain. This will involve configuring a `ChatOpenAI` LLM and integrating the Chroma vector store as a retriever to answer queries based on the mini-KB.


**Reasoning**:
To set up the RAG chain, I will import ChatOpenAI, initialize the LLM, create a retriever from the existing vector store, and then construct the RAG chain by piping the retriever and the LLM together.



```python
from langchain_openai import ChatOpenAI

# 1. Initialize a ChatOpenAI instance
llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0)

# 2. Create a retriever from the vector_store
retriever = vector_store.as_retriever()

# 3. Construct a RAG chain
rag_chain = retriever | llm

print("ChatOpenAI LLM initialized.")
print("Retriever created from vector store.")
print("RAG chain constructed.")
```

## Implement FastAPI Application

### Subtask:
Create a FastAPI application. Define a Pydantic model for the incoming requests and implement the `/coach` endpoint. This endpoint will accept a query, pass it to the RAG chain, and return the generated response.


**Reasoning**:
To implement the FastAPI application, I will define the main application logic, including importing necessary modules, initializing the FastAPI app, defining the Pydantic request model, and creating the `/coach` endpoint that utilizes the previously defined `rag_chain`.



```python
from fastapi import FastAPI
from pydantic import BaseModel

# 2. Initialize a FastAPI application instance
app = FastAPI()

# 3. Define a Pydantic model named QueryRequest
class QueryRequest(BaseModel):
    query: str

# 4 & 5. Define a POST endpoint /coach that accepts a QueryRequest object
# and uses the rag_chain to process the query
@app.post("/coach")
async def coach_endpoint(request: QueryRequest):
    try:
        # Process the query using the RAG chain
        response = rag_chain.invoke(request.query)
        # 6. Return the generated response
        return {"response": response}
    except Exception as e:
        return {"error": str(e), "message": "Failed to process query"}
```

## Expose FastAPI with Pyngrok

### Subtask:
Start the Uvicorn server to run the FastAPI application locally within Google Colab. Then, use `pyngrok` to create a public URL, allowing external access to the `/coach` API endpoint.


**Reasoning**:
To expose the FastAPI application, I will use `nest_asyncio` to allow `uvicorn` to run non-blockingly as an `asyncio` task, then use `pyngrok` to create a public tunnel to the running server, handling the ngrok authentication token securely.



```python
import uvicorn
import nest_asyncio
import asyncio
from pyngrok import ngrok
import os
import getpass

# 1. Apply nest_asyncio to allow asyncio to be nested
nest_asyncio.apply()

# --- Run Uvicorn in a separate process/thread ---
# Uvicorn needs to run in its own event loop or thread as it's a blocking call.
# Using asyncio.create_task with nest_asyncio allows it to run concurrently.
async def run_uvicorn():
    config = uvicorn.Config(app, host="0.0.0.0", port=8000, log_level="info")
    server = uvicorn.Server(config)
    await server.serve()

print("Starting Uvicorn server...")
# 2. Run the uvicorn server in a separate task to keep the notebook responsive
uvicorn_task = asyncio.create_task(run_uvicorn())

# --- Pyngrok Setup ---
# Check for NGROK_AUTH_TOKEN and prompt if not set
if "NGROK_AUTH_TOKEN" not in os.environ:
    print("To use pyngrok, you need an ngrok auth token. You can get one from https://ngrok.com/signup")
    ngrok_token = getpass.getpass("Enter your ngrok auth token: ")
    os.environ["NGROK_AUTH_TOKEN"] = ngrok_token
    print("ngrok auth token set.")
else:
    print("ngrok auth token already set.")

# 3. Initialize pyngrok to create a tunnel to the specified port
print("Creating ngrok tunnel...")
public_url = ngrok.connect(8000).public_url

# 4 & 5. Retrieve and print the public URL generated by pyngrok
print(f"FastAPI app running at: {public_url}")
print("You can now access your API at the public URL. To stop the server, interrupt this cell.")
```

```python
import uvicorn
import nest_asyncio
import asyncio
from pyngrok import ngrok
import os
import getpass

# 1. Apply nest_asyncio to allow asyncio to be nested
nest_asyncio.apply()

# --- Run Uvicorn in a separate process/thread ---
# Uvicorn needs to run in its own event loop or thread as it's a blocking call.
# Using asyncio.create_task with nest_asyncio allows it to run concurrently.
async def run_uvicorn():
    config = uvicorn.Config(app, host="0.0.0.0", port=8000, log_level="info")
    server = uvicorn.Server(config)
    await server.serve()

print("Starting Uvicorn server...")
# 2. Run the uvicorn server in a separate task to keep the notebook responsive
uvicorn_task = asyncio.create_task(run_uvicorn())

# --- Pyngrok Setup ---
# Check for NGROK_AUTH_TOKEN and prompt if not set
if "NGROK_AUTH_TOKEN" not in os.environ:
    print("To use pyngrok, you need an ngrok auth token. You can get one from https://ngrok.com/signup")
    ngrok_token = getpass.getpass("Enter your ngrok auth token: ")
    os.environ["NGROK_AUTH_TOKEN"] = ngrok_token
    ngrok.set_auth_token(ngrok_token) # Set the auth token for pyngrok
    print("ngrok auth token set.")
else:
    print("ngrok auth token already set.")
    # Ensure pyngrok uses the already set environment variable if not already configured
    ngrok.set_auth_token(os.environ["NGROK_AUTH_TOKEN"])


# 3. Initialize pyngrok to create a tunnel to the specified port
print("Creating ngrok tunnel...")
public_url = ngrok.connect(8000).public_url

# 4 & 5. Retrieve and print the public URL generated by pyngrok
print(f"FastAPI app running at: {public_url}")
print("You can now access your API at the public URL. To stop the server, interrupt this cell.")
```
